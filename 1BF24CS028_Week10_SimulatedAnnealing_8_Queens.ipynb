```python

import random
import math

N = 8

def heuristic(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def display_board(state):
    for row in range(N):
        for col in range(N):
            if state[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


def simulated_annealing(initial_state):

    current = initial_state[:]
    current_h = heuristic(current)

    temperature = 100
    cooling_rate = 0.95
    iteration = 0

    print("\nInitial State:", current)
    print("Initial Heuristic:", current_h)

    print("\n========== CALCULATION STAGES ==========")

    while current_h > 0 and temperature > 0.1:

        iteration += 1


        neighbor = current[:]

        col = random.randint(0, N - 1)
        new_row = random.randint(0, N - 1)

        while new_row == neighbor[col]:
            new_row = random.randint(0, N - 1)

        neighbor[col] = new_row

        neighbor_h = heuristic(neighbor)


        delta = neighbor_h - current_h


        if delta < 0:
            probability = 1
        else:
            probability = math.exp(-delta / temperature)


        if random.random() < probability:
            current = neighbor
            current_h = neighbor_h
            decision = "Accepted"
        else:
            decision = "Rejected"

        print(f"\nIteration: {iteration}")
        print("Temperature:", round(temperature, 4))
        print("Neighbor State:", neighbor)
        print("Neighbor Heuristic:", neighbor_h)
        print("Delta:", delta)
        print("Acceptance Probability:", round(probability, 4))
        print("Decision:", decision)
        print("Current State:", current)
        print("Current Heuristic:", current_h)


        temperature *= cooling_rate

    print("\n========== FINAL RESULT ==========")

    if current_h == 0:
        print("Goal State Reached!")
        print("Final Array:", current)
        print("Final Heuristic:", current_h)
        print("\nFinal Chessboard:")
        display_board(current)
    else:
        print("Solution not found.")
        print("Best State:", current)
        print("Heuristic:", current_h)


print("Enter 8 values (0 to 7), separated by spaces.")
print("Each value represents the row of a queen in that column.")

initial_state = list(map(int, input("Enter initial array: ").split()))

if len(initial_state) == 8 and all(0 <= x < 8 for x in initial_state):
    simulated_annealing(initial_state)
else:
    print("Invalid input! Enter exactly 8 values between 0 and 7.")
```
